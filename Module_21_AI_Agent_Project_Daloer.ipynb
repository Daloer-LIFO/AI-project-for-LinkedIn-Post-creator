**Step-1 Requirements of the Assignment**

The assignment is requiring an AI Agent using LangChain that generates professional LinkedIn posts. The user is providing:

* A topic for the post  

* A language for the post

The AI Agent is then generating a structured and engaging LinkedIn-style post in the selected language, consisting of approximately 2–4 paragraphs. The required deliverables are a Code Repository/Notebook and a 2–3 minute demonstration video.

```python
# Step 3: Adding Gemini API Key
import os
from getpass import getpass

os.environ["GOOGLE_API_KEY"] = getpass("Enter your Gemini API Key: ")
```

```python
# Step 4: Importing LangChain Components
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# These components are being used for:

# Connecting with the LLM
# Creating the prompt template
# Processing the generated output
```

```python
# Step 5: Creating the LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.7
)

# gemini-3.6-flash is being used as the language model.
# temperature=0.7 is allowing some creativity while keeping the output professional.
```

```python
# Step 6: Designing the Prompt Template
template = """
You are a professional LinkedIn content writer.

Your task is generating a professional, engaging, and well-structured
LinkedIn post.

Topic: {topic}
Language: {language}

Requirements:
- Write the entire post in the selected language.
- Create 2 to 4 paragraphs.
- Use a professional and engaging tone.
- Start with an interesting opening.
- Explain the importance of the topic clearly.
- Include useful insights.
- End with a strong conclusion or question encouraging engagement.
- Add relevant hashtags at the end.

Generate only the LinkedIn post without additional explanations.
"""
prompt = PromptTemplate.from_template(template)
```

```python
# Step 7: Connecting the Components
chain = prompt | llm | StrOutputParser()
```

# The pipeline is working as follows:

User Input  
    ↓  
Prompt Template  
    ↓  
LLM  
    ↓  
Output Parser  
    ↓  
Generated LinkedIn Post

```python
# Step 8: Defining the AI Agent Function
def generate_linkedin_post(topic, language):
    result = chain.invoke({
        "topic": topic,
        "language": language
    })

    return result

# This function is accepting:

# topic
# language

# It is then returning the generated LinkedIn post.
```

```python
# Step 9: Generating an English LinkedIn Post
topic = "AI in Healthcare"
language = "English"

post = generate_linkedin_post(topic, language)

print(post)
```
